```python
import sympy as sp
```

```python
x, y = sp.symbols('x y')
```

```python
per = (2*x + 3*y)**2 - 3*x*(3/4*x + 4*y)
```

```python
print("Исходное выражение:")
print(per)
print()
```

```python

# Упрощаем выражение
per_simplified = sp.simplify(per)
print("Упрощенное выражение (simplify):")
print(per_simplified)
print()
```
