Поскольку дата InvoiceDate может быть отличаться для разных транзакций внутри одной покупки, при группировке возьмите самую раннюю из дат. Также отсортируйте её по датам по возрастанию.

Учтите, что InvoiceDate — не уникальное поле, данные стоит дополнительно отсортировать по InvoiceNo, перед тем как брать первое значение группы.

Сколько получилось строк?

#### Решение

````python
purchases = df_clean[['InvoiceNo', 'InvoiceDate', 'Price', 'CustomerID', 'Country']].sort_values(by='InvoiceNo')
purchases
````

````python
# Сначала отсортируем данные по дате и номеру заказа
purchases_sorted = df_clean.sort_values(['InvoiceDate', 'InvoiceNo'])

# Затем сгруппируем по номеру заказа, взяв самую раннюю дату и суммируя стоимость
purchases_grouped = purchases_sorted.groupby('InvoiceNo').agg({
    'InvoiceDate': 'first',  # первая дата (самая ранняя из-за сортировки)
    'Price': 'sum',          # сумма стоимости
    'CustomerID': 'first',   # первый ID клиента (предполагается одинаковость в рамках заказа)
    'Country': 'first'       # первая страна (аналогично)
}).reset_index()

# Сортируем по дате по возрастанию
purchases_grouped = purchases_grouped.sort_values('InvoiceDate')

# Просмотрим результат
purchases_grouped.shape
````

## ⭐️⭐️⭐️ Вопрос 21

Мы запускаем в Италии рекомендательную систему "С этим товаром часто покупают...", и для этого хотим узнать, какие различные товары чаще всего встречаются в одной покупке из этой страны. Определите, какая пара различных товаров чаще всего встречается в различных покупках с ```Country=='Italy'```, и в скольких покупках это происходит. Одинаковые товары или нет, проверяйте по равенству поля Description.

(ответ: название (Description) первого товара, название (Description) второго товара, целое число)

#### Решение

````python
import pandas as pd
from itertools import combinations

# Загрузка данных (замените на ваш путь к файлу)
# df = pd.read_csv('your_file.csv')

# 1. Фильтрация заказов из Италии
italian_orders = df[df['Country'] == 'Italy']
italian_orders
````

````python
# 2. Функция для генерации пар товаров в одном заказе
def find_product_pairs(group):
    # Получаем уникальные описания товаров в заказе
    unique_products = group['Description'].unique()
    # Генерируем все возможные пары различных товаров
    return list(combinations(unique_products, 2))
    
````

````python
# 3. Группировка по номеру заказа (InvoiceNo) и генерация пар
pairs_by_order = italian_orders.groupby('InvoiceNo').apply(find_product_pairs)

# 4. Преобразование списка пар в один плоский список
all_pairs = []
for order_pairs in pairs_by_order:
    all_pairs.extend(order_pairs)
all_pairs
````

````python
# 5. Подсчет частоты каждой пары
from collections import Counter
pair_counts = Counter(all_pairs)
pair_counts
````

````python
# 6. Находим самую частую пару
if pair_counts:
    most_common_pair, count = pair_counts.most_common(1)[0]
    product1, product2 = most_common_pair
    print(f"Ответ: {product1}, {product2}, {count}")
else:
    print("В заказах из Италии не найдено пар различных товаров.")
````
