```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
#читаем из файла
sal = pd.read_csv("wage-data.csv")
#создаем серию группированную по ЗП
sal_ren = sal.groupby('State')['Salary'].agg('mean')
#сортировка по возрастанию с изменением серии
sal_ren.sort_values(inplace=True)
sal_ren
```

```python
#создание полотна для графика
plt.figure(figsize=(12, 6))
#plt.plot(sal_ren.index, sal_ren.values, marker='o', linestyle='-', linewidth=1)
#нанесение значений серии на график
plt.plot(sal_ren)
#Настройка графика
plt.title('Средняя зарплата по штатам', fontsize=14)
plt.xlabel('Штат', fontsize=12)
plt.ylabel('Средняя зарплата', fontsize=12)
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()

#Показать график
plt.show()
```
