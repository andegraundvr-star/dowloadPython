```python
# 1. Определяем ВСЕ символьные переменные
x, g, w1, w2 = sp.symbols('x g w1 w2')

# 2. Создаем символьные выражения
f_expr = 0.5 * x        # f(x) = 0.5x
a_expr = -1.5 * g       # a(g) = -1.5g
w1_expr = w1 * 8        # Просто 8*w1 (линейная функция)
w2_expr = w2 * 2        # Просто 2*w2

# 3. Вычисляем производные
df_dx = sp.diff(f_expr, x)      # по x
df_dg = sp.diff(a_expr, g)      # по g
df_dw1 = sp.diff(w1_expr, w1)   # по w1
df_dw2 = sp.diff(w2_expr, w2)   # по w2

print(f"df/dx = {df_dx}")      # 0.5
print(f"da/dg = {df_dg}")      # -1.5
print(f"d(8*w1)/dw1 = {df_dw1}")  # 8
print(f"d(2*w2)/dw2 = {df_dw2}")  # 2
```

```python
ошибка  = (-0.01355) × 0.5 = -0.006775
ошибка  = (-0.01355) × (-1.5) = 0.020325
ошибка  = (-0.01355) × 8 = -0.1084
ошибка  = (-0.01355) × 2 = -0.0271
```
