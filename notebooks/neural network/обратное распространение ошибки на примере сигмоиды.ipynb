```python
import numpy as np

class Var:
    """
    Автоматическое дифференцирование (Autograd)
    Хранит значение и градиент, строит граф вычислений
    """
    
    def __init__(self, value, grad=0.0, parents=(), op=''):
        """
        Инициализация переменной
        
        Parameters:
        -----------
        value : float
            Численное значение переменной
        grad : float
            Накопленный градиент (df/dx)
        parents : tuple of Var
            Родительские переменные в графе вычислений
        op : str
            Операция, которая создала эту переменную (для отладки)
        """
        self.value = value      # Значение переменной (например, 8.0)
        self.grad = grad        # Градиент df/dx (начинаем с 0)
        self.parents = parents  # Родительские узлы графа
        self.op = op            # Название операции ('+', '*', 'exp', etc.)
        self._backward = lambda: None  # Функция обратного распространения
    
    def backward(self, grad=1.0):
        """
        Запуск обратного распространения градиента
        
        Parameters:
        -----------
        grad : float
            Начальный градиент (обычно df/df = 1)
        """
        # 1. Строим топологический порядок графа (от листьев к корню)
        topo = []
        visited = set()
        
        def build_topo(v):
            """Рекурсивно строим топологическую сортировку"""
            if v not in visited:
                visited.add(v)
                for parent in v.parents:  # Сначала посещаем родителей
                    build_topo(parent)
                topo.append(v)  # Затем добавляем саму переменную
        
        build_topo(self)
        
        # 2. Инициализируем градиент корневой переменной
        self.grad = grad  # df/df = 1
        
        # 3. Обратный проход в обратном порядке
        for v in reversed(topo):
            v._backward()  # Вызываем функцию обратного распространения
    
    def __repr__(self):
        """Строковое представление для отладки"""
        return f"Var(value={self.value:.6f}, grad={self.grad:.6f})"
    
    # ---------- АРИФМЕТИЧЕСКИЕ ОПЕРАЦИИ ----------
    
    def __add__(self, other):
        """Сложение: self + other"""
        # Преобразуем other в Var если нужно
        other = other if isinstance(other, Var) else Var(other)
        
        # Создаем новую переменную как результат сложения
        out = Var(
            value=self.value + other.value,
            parents=(self, other),  # Родители: self и other
            op='+'
        )
        
        def _backward():
            """
            Локальное правило обратного распространения для сложения
            
            Правило: если out = a + b, то:
            d(out)/da = 1
            d(out)/db = 1
            
            По цепному правилу:
            dL/da += dL/dout * d(out)/da
            dL/db += dL/dout * d(out)/db
            """
            # Накопление градиента (используем += так как один узел может быть родителем нескольких)
            self.grad += 1.0 * out.grad      # dL/dself += dL/dout * 1
            other.grad += 1.0 * out.grad     # dL/dother += dL/dout * 1
        
        # Привязываем функцию обратного распространения
        out._backward = _backward
        
        return out
    
    def __mul__(self, other):
        """Умножение: self * other"""
        other = other if isinstance(other, Var) else Var(other)
        
        out = Var(
            value=self.value * other.value,
            parents=(self, other),
            op='*'
        )
        
        def _backward():
            """
            Правило для умножения:
            Если out = a * b, то:
            d(out)/da = b
            d(out)/db = a
            """
            self.grad += other.value * out.grad   # dL/dself += dL/dout * other.value
            other.grad += self.value * out.grad   # dL/dother += dL/dout * self.value
        
        out._backward = _backward
        return out
    
    def __neg__(self):
        """Унарный минус: -self"""
        return self * Var(-1.0)  # Умножение на -1
    
    def __sub__(self, other):
        """Вычитание: self - other"""
        return self + (-other)  # a - b = a + (-b)
    
    def __truediv__(self, other):
        """Деление: self / other"""
        other = other if isinstance(other, Var) else Var(other)
        
        out = Var(
            value=self.value / other.value,
            parents=(self, other),
            op='/'
        )
        
        def _backward():
            """
            Правило для деления:
            Если out = a / b, то:
            d(out)/da = 1/b
            d(out)/db = -a/b²
            """
            self.grad += (1.0 / other.value) * out.grad
            other.grad += (-self.value / (other.value ** 2)) * out.grad
        
        out._backward = _backward
        return out
    
    def exp(self):
        """Экспонента: exp(self)"""
        out = Var(
            value=np.exp(self.value),
            parents=(self,),  # Родитель только self
            op='exp'
        )
        
        def _backward():
            """
            Правило для экспоненты:
            Если out = exp(a), то:
            d(out)/da = exp(a) = out.value
            """
            self.grad += out.value * out.grad  # dL/dself += dL/dout * exp(self)
        
        out._backward = _backward
        return out
    
    def sigmoid(self):
        """Сигмоида: 1/(1 + exp(-self))"""
        # Реализуем через существующие операции
        neg = -self
        exp_neg = neg.exp()
        one_plus = Var(1.0) + exp_neg
        return Var(1.0) / one_plus
```

```python
x = Var(8.0)
y = Var(2.0)
w1 = Var(0.5)
w2 = Var(-1.5)

# ПРЯМОЙ ПРОХОД (автоматически строит граф вычислений)
z = w1 * x + w2 * y    # z = 0.5*8 + (-1.5)*2 = 1.0
s = -z                 # s = -1.0
t = s.exp()            # t = exp(-1.0) = 0.367879
u = Var(1.0) + t       # u = 1 + 0.367879 = 1.367879
f = Var(1.0) / u       # f = 1/1.367879 = 0.731058

print("Прямой проход:")
print(f"z = {z.value:.6f}, s = {s.value:.6f}, t = {t.value:.6f}, "
      f"u = {u.value:.6f}, f = {f.value:.6f}")

# ОБРАТНЫЙ ПРОХОД (автоматически вычисляет все градиенты)
# Устанавливаем градиент целевой функции
# Предположим, наша функция потерь L = f, и dL/df = -0.068941
f.grad = -0.068941  # E из вашего примера

# Запускаем обратное распространение
f.backward()

print("\nОбратный проход (автоматический):")
print(f"df/dw1 = {w1.grad:.6f} (ожидалось: -0.1084)")
print(f"df/dw2 = {w2.grad:.6f} (ожидалось: -0.0271)")
print(f"df/dx  = {x.grad:.6f}  (ожидалось: -0.006775)")
print(f"df/dy  = {y.grad:.6f}  (ожидалось: 0.020325)")

# Проверка промежуточных градиентов
print(f"\nПромежуточные градиенты:")
print(f"df/du  = {u.grad:.6f}  (ожидалось: 0.03684)")
print(f"df/dt  = {t.grad:.6f}  (ожидалось: 0.03684)")
print(f"df/ds  = {s.grad:.6f}  (ожидалось: 0.01355)")
print(f"df/dz  = {z.grad:.6f}  (ожидалось: -0.01355)")
```

```python
def print_computation_graph(var, indent=0):
    """Печатает граф вычислений"""
    prefix = "  " * indent
    print(f"{prefix}{var.op}: value={var.value:.6f}, grad={var.grad:.6f}")
    for parent in var.parents:
        print_computation_graph(parent, indent + 1)

print("\nГраф вычислений (от f вниз):")
print_computation_graph(f)
```

```python
import numpy as np

# Ваши параметры
x, y = 8.0, 2.0
w1, w2 = 0.5, -1.5

# Прямой проход (сохраняем все значения для обратного)
z = w1*x + w2*y          # = 1.0
s = -z                   # = -1.0
t = np.exp(s)            # = 0.367879
u = 1 + t                # = 1.367879
f = 1 / u                # = 0.731058

# Градиент от верхнего слоя
df = -0.068941  # E

# Обратный проход (цепное правило в компактной форме)
# f = 1/u, u = 1 + t, t = exp(s), s = -z, z = w1*x + w2*y

# Вычисляем все производные сразу
df_du = -1/(u**2)        # df/du = -1/u²
df_dt = df_du * 1        # df/dt = df/du * du/dt
df_ds = df_dt * t        # df/ds = df/dt * dt/ds = df/dt * t
df_dz = df_ds * (-1)     # df/dz = df/ds * ds/dz
df_dw1 = df_dz * x       # df/dw1 = df/dz * dz/dw1
df_dw2 = df_dz * y       # df/dw2 = df/dz * dz/dw2
df_dx = df_dz * w1       # df/dx = df/dz * dz/dx
df_dy = df_dz * w2       # df/dy = df/dz * dz/dy

# Умножаем на градиент от верхнего слоя
grad_w1 = df * df_dw1
grad_w2 = df * df_dw2
grad_x = df * df_dx
grad_y = df * df_dy

print("Компактный обратный проход:")
print(f"grad_w1 = {grad_w1:.6f}")
print(f"grad_w2 = {grad_w2:.6f}")
print(f"grad_x  = {grad_x:.6f}")
print(f"grad_y  = {grad_y:.6f}")
```
