```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('wage-data-coast.csv')
df
```

```python
plt.subplots()
```

```python
fig, ax = plt.subplots()
```

```python
fig, ax = plt.subplots()
data = df[df['State'] =='Washington'][['Year', 'Salary']]
ax.plot(data['Year'], data['Salary'])
```

```python
fig, ax = plt.subplots()
data = df[df['State'] == 'Washington'][['Year', 'Salary']]

ax.plot(data['Year'], data['Salary'])
ax.set_title('Минимальная зарплата в штате Вашингтон по годам', pad = 16, color = 'navy', backgroundcolor = 'lightgray')
ax.set_xlabel('Год')
ax.set_ylabel('Минимальная зарплата, $/час')
ax.set_xticks(list(range(data['Year'].min(), data['Year'].max(),5))+[data['Year'].max()])
plt.show()
```

```python
fig, ax = plt.subplots()
data = df[(df['State'] =='Arizona') & (df['Year'] <= 2000)][['Year', 'Salary']].groupby('Year')['Salary'].min().sort_index() #сортировка по годам
data

ax.plot(data.index, data.values)
ax.set_title('Минимальная зарплата в штате Аризона до 2000 года', pad = 16, color = 'navy', backgroundcolor = 'lightgray')
ax.set_xlabel('Год', color='navy')
ax.set_ylabel('Минимальная зарплата, $/час', color='navy')
#Устанавливаем метки на оси X (каждые 5 лет)
years = sorted(data.index)  # Получаем годы из индекса
ax.set_xticks(years[::2])   # Каждые 2 года для читаемости
plt.xticks(rotation=45)
plt.show()
```
