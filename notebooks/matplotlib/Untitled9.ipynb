```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('wage-data-coast.csv')
df
data = df[df['State'] == 'Alaska']['Salary']
data1 = df[df['State'] == 'California']['Salary']
data2 = df[df['State'] == 'Washington']['Salary']
fig, axs = plt.subplots(figsize = (12,6), ncols = 1, nrows = 3)
n0, bins, patches0 = axs[0].hist(data, bins=15)
n1, bins1, patches1 = axs[1].hist(data1, bins=15)
n2, bins2, patches2 = axs[2].hist(data2, bins=15)

axs[0].set_xticks(bins)
axs[1].set_xticks(bins1)
axs[2].set_xticks(bins2)
axs[0].set_title('Распределение минимальных зарплат разных штатах')
axs[0].set_ylabel('# записей')
axs[1].set_ylabel('# записей')
axs[2].set_xlabel('Минимальная зарплата, $/час')
axs[2].set_ylabel('# записей')
for bin_ in list(bins2):
  for ax in axs:
    ax.axvline(x = bin_, color = 'lightgrey', linewidth = 1, linestyle = '--')
plt.tight_layout() 
plt.show()
```
