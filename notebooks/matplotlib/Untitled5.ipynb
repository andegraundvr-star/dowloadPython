```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#читаем файл
df_bar = pd.read_csv('wage-data-coast.csv')
df_bar
```

```python
#строим серию с выбором 2000 года ,с сортировкой Salary по убыванию. Иоставляем первые 5 строк
series1 = df_bar[df_bar['Year'] == 2000].sort_values('Salary', ascending=False).head(5)
series1
```

```python
#построение barchart
plt.bar(series1['State'], series1['Salary'])
plt.title("Наивысшее значение минимальной зарплаты в 5-ти штатах в 2000 году")
plt.ylabel('$/час')
plt.xlabel('Назвение штата')
```
