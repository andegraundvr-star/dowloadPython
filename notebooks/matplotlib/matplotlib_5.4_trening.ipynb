```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#читаем файл
df_point = pd.read_csv('wage-data-coast.csv')
df_point
```

```python
#строим серию с выборкой по 2000 году и показателю ISCoastal равному 1 
series = df_point[(df_point['Year'] == 2000) & (df_point['IsCoastal'] == 1)]['Salary']
plt.hist(series)
```

```python
series.max()
```

```python
bind_x = series.max() / 10
x_range = [i * bind_x for i in range(11)]
plt.hist(series, bins = 10)
plt.title("Гистограмма ЗП в прибрежных штатах за 2000 год")
plt.xlabel("$ в час")
plt.ylabel("Количество подствержденных фактов")
plt.xticks(x_range)
```

```python
#точечный график
plt.scatter(df_point['Year'], df_point['Salary'], c = df_point['IsCoastal'])
plt.title("Минимальная зарплата по годам")
plt.ylabel('$/час')
plt.xlabel('Год')
```

```python
#добавление легенды
df_0 = df_point[df_point['IsCoastal']== 0]
plt.scatter(df_point['Year'], df_point['Salary'], label=f"Not")
df_1 = df_point[df_point['IsCoastal']== 1]
plt.scatter(df_point['Year'], df_point['Salary'], label=f"Yes")

plt.title("Минимальная зарплата по годам")
plt.ylabel('$/час')
plt.xlabel('Год')
plt.legend()
```
