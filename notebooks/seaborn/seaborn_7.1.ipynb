```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('wage-data-coast.csv')
df
```

```python
data = df[df['Year'].isin([2000, 2010])]
ax = sns.histplot(data = data, x = 'Salary', hue = 'Year')
ax.set_title('Распределение минимальных зарплат в 2000 и 2010 годах')
ax.set_xlabel('Минимальная зарплата, $/час')
ax.set_ylabel('# записей')
```

```python
data = df[df['State'].isin(['Alaska', 'Washington', 'California', 'Virginia', 'Arizona'])]
fig, ax = plt.subplots(figsize=(12,6))
sns.lineplot(data=data, x = 'Year', y = 'Salary', hue = 'State', ax = ax)
ax.set_title('Диаграмма минимальной зарплаты в пяти штатах')
ax.set_xlabel('год')
ax.set_ylabel('Минимальная зарплата, $/час')
```

```python
data1 = df[df['Year'].isin([2000])]
ax = sns.histplot(data = data1, x = 'Salary', hue = 'IsCoastal', alpha = 0.3, bins = 25)
ax.set_title('Распределение минимальных зарплат в 2000 году')
ax.set_xlabel('Минимальная зарплата, $/час')
ax.set_ylabel('# записей')
```

```python
data1 = df[df['Year'].isin([2000])]
ax = sns.histplot(data = data1, x = 'Salary', hue = 'IsCoastal', alpha = 0.3, bins = 25, multiple="dodge")
ax.set_title('Распределение минимальных зарплат в 2000 году')
ax.set_xlabel('Минимальная зарплата, $/час')
ax.set_ylabel('# записей')
```

```python
data = df[df['State'].isin(['Alaska', 'Washington', 'California', 'Virginia', 'Arizona'])]
fig, ax = plt.subplots(figsize=(12,6))
sns.lineplot(data=data, x = 'Year', y = 'Salary', hue = 'State', ax = ax, style = 'IsCoastal')
ax.set_title('Диаграмма минимальной зарплаты в пяти штатах')
ax.set_xlabel('год')
ax.set_ylabel('Минимальная зарплата, $/час')
```

```python
df = pd.read_csv('wage-data-coast-with-population.csv')
df
```

```python
data = df[df['State'].isin(['Alaska', 'Washington', 'California', 'Virginia', 'Arizona'])]
data_t = data[data['Year'] == 2017][['State', 'Population']].sort_values('Population')
data_t.loc[:, 'Size'] = list(range(1, data_t.shape[0] + 1))#добавление солбца с показателем 1-5 в соответсвии с численностью населения
data2 = data.merge(data_t[['State', 'Size']], how = 'left', on = 'State')
data2
```

```python
data = df[df['State'].isin(['Alaska', 'Washington', 'California', 'Virginia', 'Arizona'])]
fig, ax = plt.subplots(figsize=(12,6))
sns.lineplot(data=data2, x = 'Year', y = 'Salary', hue = 'State', ax = ax, style = 'IsCoastal', size = 'Size')
ax.set_title('Диаграмма минимальной зарплаты в пяти штатах')
ax.set_xlabel('год')
ax.set_ylabel('Минимальная зарплата, $/час')
```

```python
data = df
fig, ax = plt.subplots(figsize=(12,6))
sns.scatterplot(data=data, x = 'Year', y = 'Salary',  ax = ax,  hue = 'Population', alpha = 0.8, size = 'Population')
ax.set_title('Диаграмма минимальной зарплаты в штатах США')
ax.set_xlabel('год')
ax.set_ylabel('Минимальная зарплата, $/час')
```

```python
data = df[df['Year'].isin([2000, 2015])]
fig, ax = plt.subplots(figsize=(12,6))
sns.barplot(data=data, x = 'IsCoastal', y = 'Salary', ax = ax, errorbar=None, hue = 'Year')
ax.set_title('Распределение минимальных зарплат в 2000 году')
ax.set_xlabel('Минимальная зарплата, $/час')
ax.set_ylabel('# записей')
plt.show()
```

```python
data = df
fig, ax = plt.subplots(figsize=(12,6))
sns.scatterplot(data=df, x = 'Year', y = 'Population',  ax = ax,  hue = 'Salary', alpha = 0.8, size = 'Salary')
ax.set_title('Диаграмма минимальной зарплаты в штатах США')
ax.set_xlabel('год')
ax.set_ylabel('Население')
```
