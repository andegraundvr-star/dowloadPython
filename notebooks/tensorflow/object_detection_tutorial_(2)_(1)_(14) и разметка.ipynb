By default we use an "SSD with Mobilenet" model here. See the [detection model zoo](https://github.com/tensorflow/models/blob/master/research/object_detection/g3doc/detection_model_zoo.md) for a list of other models that can be run out-of-the-box with varying speeds and accuracies.

## Loader

```
def load_model(model_name):
  base_url = 'http://download.tensorflow.org/models/object_detection/'
  model_file = model_name + '.tar.gz'
  model_dir = tf.keras.utils.get_file(
    fname=model_name,
    origin=base_url + model_file,
    untar=True)

  model_dir = pathlib.Path(model_dir)/"saved_model"

  model = tf.saved_model.load(str(model_dir))

  return model
```

Add a wrapper function to call the model, and cleanup the outputs:

```
def run_inference_for_single_image(model, image):
  image = np.asarray(image)
  # The input needs to be a tensor, convert it using `tf.convert_to_tensor`.
  input_tensor = tf.convert_to_tensor(image)
  # The model expects a batch of images, so add an axis with `tf.newaxis`.
  input_tensor = input_tensor[tf.newaxis,...]

  # Run inference
  model_fn = model.signatures['serving_default']
  output_dict = model_fn(input_tensor)

  # All outputs are batches tensors.
  # Convert to numpy arrays, and take index [0] to remove the batch dimension.
  # We're only interested in the first num_detections.
  num_detections = int(output_dict.pop('num_detections'))
  output_dict = {key:value[0, :num_detections].numpy()
                 for key,value in output_dict.items()}
  output_dict['num_detections'] = num_detections

  # detection_classes should be ints.
  output_dict['detection_classes'] = output_dict['detection_classes'].astype(np.int64)

  # Handle models with masks:
  if 'detection_masks' in output_dict:
    # Reframe the the bbox mask to the image size.
    detection_masks_reframed = utils_ops.reframe_box_masks_to_image_masks(
              output_dict['detection_masks'], output_dict['detection_boxes'],
               image.shape[0], image.shape[1])
    detection_masks_reframed = tf.cast(detection_masks_reframed > 0.5,
                                       tf.uint8)
    output_dict['detection_masks_reframed'] = detection_masks_reframed.numpy()

  return output_dict
```

Run it on each test image and show the results:

```
def show_inference(model, image_path):
  # the array based representation of the image will be used later in order to prepare the
  # result image with boxes and labels on it.
  image_np = np.array(Image.open(image_path))
  # Actual detection.
  output_dict = run_inference_for_single_image(model, image_np)
  # Visualization of the results of a detection.
  vis_util.visualize_boxes_and_labels_on_image_array(
      image_np,
      output_dict['detection_boxes'],
      output_dict['detection_classes'],
      output_dict['detection_scores'],
      category_index,
      instance_masks=output_dict.get('detection_masks_reframed', None),
      use_normalized_coordinates=True,
      line_thickness=8)

  display(Image.fromarray(image_np))
```

```
%%time
import tensorflow as tf
from object_detection.utils import dataset_util
from PIL import Image
import os
import pandas as pd
import csv


def create_tf_example(photo_id, photo_info):
    """Создает TF Example для одного фото"""
    photo_id = photo_data['id']
    base_path = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data'
    img_fpath = os.path.join(base_path, photo_id)

    #открываем изображение для получения размеров
    try:
        img = Image.open(img_fpath)
        width, height = img.size
    except:
        #если не получается, используем стандартные размеры
        width, height = 480, 640

    filename = str.encode(photo_id)

    with open(img_fpath, mode='rb') as f:
        encoded_image_data = f.read()
    image_format = b'jpeg'

    #нормализуем координаты
    xmins_norm = [xmin / float(width) for xmin in photo_info['xmins']]
    xmaxs_norm = [xmax / float(width) for xmax in photo_info['xmaxs']]
    ymins_norm = [ymin / float(height) for ymin in photo_info['ymins']]
    ymaxs_norm = [ymax / float(height) for ymax in photo_info['ymaxs']]

    #конвертируем названия классов в bytes
    classes_text_bytes = [text.encode() for text in photo_info['classes_text']]

    tf_example = tf.train.Example(features=tf.train.Features(feature={
        'image/height': dataset_util.int64_feature(height),
        'image/width': dataset_util.int64_feature(width),
        'image/filename': dataset_util.bytes_feature(filename),
        'image/source_id': dataset_util.bytes_feature(filename),
        'image/encoded': dataset_util.bytes_feature(encoded_image_data),
        'image/format': dataset_util.bytes_feature(image_format),
        'image/object/bbox/xmin': dataset_util.float_list_feature(xmins_norm),
        'image/object/bbox/xmax': dataset_util.float_list_feature(xmaxs_norm),
        'image/object/bbox/ymin': dataset_util.float_list_feature(ymins_norm),
        'image/object/bbox/ymax': dataset_util.float_list_feature(ymaxs_norm),
        'image/object/class/text': dataset_util.bytes_list_feature(classes_text_bytes),
        'image/object/class/label': dataset_util.int64_list_feature(photo_info['classes']),
    }))
    return tf_example
```

```
import pandas as pd
from pathlib import Path
import numpy as np

DATA_PATH = Path("/content/drive/MyDrive/object_detection_project/data")
print("=== СРОЧНАЯ ПРОВЕРКА ДАННЫХ ===")

# 1. Проверяем CSV
print("1. Проверяем annotations.csv:")
df = pd.read_csv(DATA_PATH / "annotations.csv")

# Находим ВСЕ ошибки
errors = []
for idx, row in df.iterrows():
    if row['xmin'] > row['xmax']:
        errors.append((idx, row['id'], 'xmin > xmax', f"{row['xmin']} > {row['xmax']}"))
    if row['ymin'] > row['ymax']:
        errors.append((idx, row['id'], 'ymin > ymax', f"{row['ymin']} > {row['ymax']}"))

print(f"   Найдено ошибок: {len(errors)}")
for idx, file_id, error_type, details in errors[:5]:  # Показываем первые 5
    print(f"   Строка {idx}: {file_id} - {error_type} ({details})")

# 2. Создаём ИДЕАЛЬНО исправленный CSV
print("\n2. Создаём идеально исправленный CSV...")
df_perfect = df.copy()

# Автоматически исправляем ВСЕ ошибки
fixed_count = 0
for idx, row in df_perfect.iterrows():
    if row['xmin'] > row['xmax']:
        df_perfect.at[idx, 'xmin'], df_perfect.at[idx, 'xmax'] = row['xmax'], row['xmin']
        fixed_count += 1
    if row['ymin'] > row['ymax']:
        df_perfect.at[idx, 'ymin'], df_perfect.at[idx, 'ymax'] = row['ymax'], row['ymin']
        fixed_count += 1

print(f"   Исправлено ошибок: {fixed_count}")

# Сохраняем
perfect_csv = DATA_PATH / "annotations_PERFECT_FINAL.csv"
df_perfect.to_csv(perfect_csv, index=False)
print(f"   Сохранён: {perfect_csv.name}")

# 3. Проверяем координаты после исправления
print("\n3. Проверка после исправления:")
x_errors = len(df_perfect[df_perfect['xmin'] > df_perfect['xmax']])
y_errors = len(df_perfect[df_perfect['ymin'] > df_perfect['ymax']])
print(f"   Осталось ошибок xmin > xmax: {x_errors}")
print(f"   Осталось ошибок ymin > ymax: {y_errors}")

if x_errors == 0 and y_errors == 0:
    print("   ✅ Координаты исправлены!")
else:
    print("   ❌ Всё ещё есть ошибки!")
```

```
# Создаём абсолютно правильный TFRecord
create_perfect_tfrecord = '''
import tensorflow as tf
import pandas as pd
from pathlib import Path
import io
from PIL import Image
import sys

print("=== СОЗДАНИЕ АБСОЛЮТНО ИСПРАВЛЕННОГО TFRecord ===")

DATA_PATH = Path("/content/drive/MyDrive/object_detection_project/data")
df = pd.read_csv(DATA_PATH / "annotations_PERFECT_FINAL.csv")
df = df.rename(columns={"id": "filename"})

print(f"Файлов: {len(df['filename'].unique())}")
print(f"Объектов: {len(df)}")

# ДОПОЛНИТЕЛЬНАЯ проверка
print("\\nДополнительная проверка координат:")
for idx, row in df.iterrows():
    if row['xmin'] > row['xmax']:
        print(f"  ⚠️  Строка {idx}: {row['filename']} - xmin={row['xmin']} > xmax={row['xmax']}")
    if row['ymin'] > row['ymax']:
        print(f"  ⚠️  Строка {idx}: {row['filename']} - ymin={row['ymin']} > ymax={row['ymax']}")

# API
sys.path.append("/content/models/research")
sys.path.append("/content/models/research/slim")
from object_detection.utils import dataset_util

def create_example_with_force_fix(filename, group_df):
    """Создаёт пример с ПРИНУДИТЕЛЬНЫМ исправлением координат"""
    img_path = DATA_PATH / filename

    # Читаем изображение
    with tf.io.gfile.GFile(str(img_path), "rb") as fid:
        encoded_image = fid.read()

    # Размеры
    image = Image.open(io.BytesIO(encoded_image))
    width, height = image.size

    # Координаты с ПРИНУДИТЕЛЬНЫМ исправлением
    xmins, xmaxs, ymins, ymaxs = [], [], [], []

    for _, obj in group_df.iterrows():
        # Нормализуем
        xmin = obj["xmin"] / width
        xmax = obj["xmax"] / width
        ymin = obj["ymin"] / height
        ymax = obj["ymax"] / height

        # ПРИНУДИТЕЛЬНОЕ исправление
        if xmin > xmax:
            print(f"  ⚠️  ФОРСИРОВАННОЕ исправление: {filename} xmin={xmin} > xmax={xmax}")
            xmin, xmax = xmax, xmin

        if ymin > ymax:
            print(f"  ⚠️  ФОРСИРОВАННОЕ исправление: {filename} ymin={ymin} > ymax={ymax}")
            ymin, ymax = ymax, ymin

        # Гарантируем диапазон
        xmin = max(0.0, min(1.0, xmin))
        xmax = max(0.0, min(1.0, xmax))
        ymin = max(0.0, min(1.0, ymin))
        ymax = max(0.0, min(1.0, ymax))

        # Гарантируем минимальный размер
        if xmax - xmin < 0.01:
            xmax = xmin + 0.01
        if ymax - ymin < 0.01:
            ymax = ymin + 0.01

        xmins.append(xmin)
        xmaxs.append(xmax)
        ymins.append(ymin)
        ymaxs.append(ymax)

    classes = ["doll", "dwarf"]
    classes_text = [c.encode("utf8") for c in group_df["class_name"]]
    class_ids = [classes.index(c) + 1 for c in group_df["class_name"]]

    return tf.train.Example(features=tf.train.Features(feature={
        "image/height": dataset_util.int64_feature(height),
        "image/width": dataset_util.int64_feature(width),
        "image/encoded": dataset_util.bytes_feature(encoded_image),
        "image/object/bbox/xmin": dataset_util.float_list_feature(xmins),
        "image/object/bbox/xmax": dataset_util.float_list_feature(xmaxs),
        "image/object/bbox/ymin": dataset_util.float_list_feature(ymins),
        "image/object/bbox/ymax": dataset_util.float_list_feature(ymaxs),
        "image/object/class/text": dataset_util.bytes_list_feature(classes_text),
        "image/object/class/label": dataset_util.int64_list_feature(class_ids),
    }))

# Создаём
print("\\nСоздание TFRecord...")
all_files = df["filename"].unique()
output_path = DATA_PATH / "train_ABSOLUTELY_PERFECT.record"

with tf.io.TFRecordWriter(str(output_path)) as writer:
    for i, filename in enumerate(all_files, 1):
        group_df = df[df["filename"] == filename]
        example = create_example_with_force_fix(filename, group_df)
        writer.write(example.SerializeToString())

        if i % 10 == 0:
            print(f"  Обработано {i}/{len(all_files)} файлов")

print(f"\\n✅ Создан {output_path.name}")
print(f"   Размер: {output_path.stat().st_size / (1024*1024):.1f} MB")
'''

# Запускаем создание
with open('/content/create_absolutely_perfect.py', 'w') as f:
    f.write(create_perfect_tfrecord)

print("\nСоздаём АБСОЛЮТНО исправленный TFRecord...")
!/content/tf_od_venv_py39/bin/python /content/create_absolutely_perfect.py
```

```
print("🔧 АДАПТАЦИЯ КОДА ДЛЯ НОВОЙ МОДЕЛИ")
print("="*60)

# 1. Импортируем все необходимое
import tensorflow as tf
import numpy as np
import os
import matplotlib.pyplot as plt
from PIL import Image
import imageio
import sys

# Добавляем пути к object_detection
sys.path.append('/content/models/research')
sys.path.append('/content/models/research/slim')

print("1. Импортируем модули object_detection...")
try:
    from object_detection.utils import label_map_util
    from object_detection.utils import visualization_utils as vis_util
    print("✅ Модули object_detection загружены")
except ImportError as e:
    print(f"❌ Ошибка импорта: {e}")
    print("Пробуем установить...")
    !source /content/tf_od_venv_py39/bin/activate && cd /content/models/research && pip install -q .
    from object_detection.utils import label_map_util
    from object_detection.utils import visualization_utils as vis_util

# 2. Загружаем модель
print("\n2. Загружаем экспортированную модель...")
MODEL_PATH = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/exported_model/saved_model'

# Проверяем разные возможные пути
if not os.path.exists(MODEL_PATH):
    # Пробуем найти модель
    possible_paths = [
        '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/exported_model',
        '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/exported_model/saved_model',
        '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/output_test'
    ]

    for path in possible_paths:
        if os.path.exists(path):
            MODEL_PATH = path
            print(f"✅ Найдена модель: {path}")
            break

try:
    # Загружаем SavedModel
    print(f"Загрузка модели из {MODEL_PATH}...")
    model = tf.saved_model.load(MODEL_PATH)
    detect_fn = model.signatures['serving_default']
    print("✅ Модель загружена успешно!")
    print(f"Сигнатуры модели: {list(model.signatures.keys())}")
except Exception as e:
    print(f"❌ Ошибка загрузки модели: {e}")
    print("\nСначала экспортируем модель...")

    # Экспортируем модель
    export_cmd = f'''
    source /content/tf_od_venv_py39/bin/activate
    cd /content/models/research && \
    python object_detection/exporter_main_v2.py \
        --input_type image_tensor \
        --pipeline_config_path "/content/drive/My Drive/tf_od_demo/my_data_2/my_data/pipeline.config" \
        --trained_checkpoint_dir "/content/drive/My Drive/tf_od_demo/my_data_2/my_data/output_test" \
        --output_directory "/content/drive/My Drive/tf_od_demo/my_data_2/my_data/exported_model" \
        --alsologtostderr
    '''

    print("Экспортируем модель...")
    !{export_cmd}

    # Пробуем снова
    MODEL_PATH = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/exported_model/saved_model'
    model = tf.saved_model.load(MODEL_PATH)
    detect_fn = model.signatures['serving_default']
    print("✅ Модель экспортирована и загружена!")

# 3. Загружаем label map (для вашей новой модели с 2 классами)
print("\n3. Загружаем label map...")
LABEL_MAP_PATH = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/label_map.pbtxt'

if not os.path.exists(LABEL_MAP_PATH):
    print(f"❌ Файл label_map не найден: {LABEL_MAP_PATH}")
    print("Создаем новый label_map для 2 классов...")

    label_content = '''item {
  id: 1
  name: 'object1'
}
item {
  id: 2
  name: 'object2'
}'''

    with open(LABEL_MAP_PATH, 'w') as f:
        f.write(label_content)
    print(f"✅ label_map создан: {LABEL_MAP_PATH}")

try:
    category_index = label_map_util.create_category_index_from_labelmap(
        LABEL_MAP_PATH,
        use_display_name=True
    )
    print(f"✅ Категории загружены: {category_index}")
except Exception as e:
    print(f"❌ Ошибка загрузки label map: {e}")
    category_index = {1: {'id': 1, 'name': 'object1'},
                      2: {'id': 2, 'name': 'object2'}}
    print(f"Используем категории по умолчанию: {category_index}")

# 4. Функция детекции для TF2 (аналогично вашей)
def detect_objects(image_np, model):
    """Детекция объектов на одном изображении"""
    # Конвертируем в тензор TF
    input_tensor = tf.convert_to_tensor(image_np)
    # Добавляем batch dimension (1, height, width, 3)
    input_tensor = input_tensor[tf.newaxis, ...]

    # Выполняем детекцию
    detections = model(input_tensor)

    # Обрабатываем результаты
    num_detections = int(detections.pop('num_detections'))

    # Конвертируем в numpy
    result = {}
    for key, value in detections.items():
        result[key] = value[0, :num_detections].numpy()

    result['num_detections'] = num_detections
    result['detection_classes'] = result['detection_classes'].astype(np.int64)

    return result

# 5. Загружаем тестовое изображение
print("\n4. Загружаем тестовое изображение...")
IMAGE_PATH = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/test.jpg'

# Проверяем разные пути
if not os.path.exists(IMAGE_PATH):
    print(f"❌ Файл не найден: {IMAGE_PATH}")
    print("\nИщем изображения в папке...")

    data_folder = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data'
    !find "{data_folder}" -name "*.jpg" -o -name "*.jpeg" -o -name "*.png" -o -name "*.JPG" 2>/dev/null | head -10

    # Берем первое найденное изображение
    import glob
    images = glob.glob(f"{data_folder}/*.jpg") + glob.glob(f"{data_folder}/*.jpeg") + \
             glob.glob(f"{data_folder}/*.png") + glob.glob(f"{data_folder}/*.JPG")

    if images:
        IMAGE_PATH = images[0]
        print(f"✅ Используем первое найденное изображение: {IMAGE_PATH}")
    else:
        # Создаем тестовое изображение
        print("Создаем тестовое изображение...")
        IMAGE_PATH = os.path.join(data_folder, 'test_image.jpg')
        test_image = np.zeros((640, 640, 3), dtype=np.uint8)
        test_image[200:400, 150:350] = [255, 100, 100]   # Объект 1
        test_image[300:500, 400:550] = [100, 255, 100]   # Объект 2

        Image.fromarray(test_image).save(IMAGE_PATH)
        print(f"✅ Создано тестовое изображение: {IMAGE_PATH}")

# Загружаем изображение
print(f"Загружаем изображение: {IMAGE_PATH}")
try:
    image_np = imageio.imread(IMAGE_PATH)
    print(f"✅ Изображение загружено. Размер: {image_np.shape}")

    # Если у изображения есть альфа-канал, убираем его
    if image_np.shape[-1] == 4:
        print("Убираем альфа-канал...")
        image_np = image_np[..., :3]

    # Если изображение не 640x640, меняем размер (как в модели)
    if image_np.shape[:2] != (640, 640):
        print(f"Меняем размер с {image_np.shape[:2]} на (640, 640)...")
        image_pil = Image.fromarray(image_np)
        image_pil = image_pil.resize((640, 640))
        image_np = np.array(image_pil)
        print(f"Новый размер: {image_np.shape}")

except Exception as e:
    print(f"❌ Ошибка загрузки изображения: {e}")
    exit(1)

# 6. Запускаем детекцию
print("\n5. Запускаем детекцию...")
try:
    output_dict = detect_objects(image_np, detect_fn)
    print(f"✅ Детекция завершена. Всего обнаружений: {output_dict['num_detections']}")
except Exception as e:
    print(f"❌ Ошибка детекции: {e}")
    exit(1)

# 7. Анализируем результаты
confidence_threshold = 0.3

# Сколько уверенных детекций
confident_mask = output_dict['detection_scores'] > confidence_threshold
confident_count = np.sum(confident_mask)

print(f"\n6. Анализ результатов:")
print(f"  Порог уверенности: {confidence_threshold}")
print(f"  Уверенных детекций: {confident_count}")

if confident_count > 0:
    print(f"  Оценки уверенности: {output_dict['detection_scores'][confident_mask]}")
    print(f"  Классы: {output_dict['detection_classes'][confident_mask]}")

    # Показываем информацию по каждому обнаружению
    print(f"\n  Детали обнаружений:")
    for i in range(min(confident_count, 5)):  # Показываем первые 5
        class_id = output_dict['detection_classes'][confident_mask][i]
        score = output_dict['detection_scores'][confident_mask][i]
        class_name = category_index.get(class_id, {}).get('name', f'class_{class_id}')
        print(f"    {i+1}. {class_name}: уверенность {score:.2%}")
else:
    print("  Нет уверенных детекций. Попробуйте уменьшить порог.")

# 8. Визуализация
print("\n7. Визуализация результатов...")

# Создаем копию изображения для рисования
image_with_boxes = image_np.copy()

# Рисуем bounding boxes
vis_util.visualize_boxes_and_labels_on_image_array(
    image_with_boxes,
    output_dict['detection_boxes'],
    output_dict['detection_classes'],
    output_dict['detection_scores'],
    category_index,
    use_normalized_coordinates=True,
    line_thickness=4,
    min_score_thresh=confidence_threshold,
    max_boxes_to_draw=20)

# Сохраняем результат
output_image_path = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/detection_result.jpg'
Image.fromarray(image_with_boxes).save(output_image_path)
print(f"✅ Результат сохранен: {output_image_path}")

# 9. Отображаем результат
plt.figure(figsize=(16, 12))
plt.imshow(image_with_boxes)
plt.axis('off')

if confident_count > 0:
    plt.title(f'Обнаружено объектов: {confident_count} (порог: {confidence_threshold})',
              fontsize=16, color='green')
else:
    plt.title(f'Объекты не обнаружены (порог: {confidence_threshold})',
              fontsize=16, color='red')

plt.tight_layout()

# Сохраняем в файл для отображения
temp_path = '/content/temp_result.png'
plt.savefig(temp_path, bbox_inches='tight', pad_inches=0, dpi=100)
plt.close()

# Показываем результат в Colab
from IPython.display import Image as IPImage, display
display(IPImage(filename=temp_path))

print(f"\n8. Итог:")
print(f"  Изображение: {IMAGE_PATH}")
print(f"  Модель: {MODEL_PATH}")
print(f"  Найдено объектов: {confident_count}")
print(f"  Результат сохранен: {output_image_path}")

# 10. Если нужно, тестируем с другим порогом
print("\n9. Тестирование с разными порогами...")
thresholds = [0.1, 0.2, 0.3, 0.5, 0.7]

print(f"  Порог | Найдено объектов")
print(f"  {'-'*23}")
for thresh in thresholds:
    count = np.sum(output_dict['detection_scores'] > thresh)
    print(f"  {thresh:.1f}   | {count}")

# 11. Дополнительная информация о модели
print("\n10. Информация о модели:")
print(f"  Входной тензор: [batch, height, width, channels]")
print(f"  Ожидаемый размер: (1, 640, 640, 3)")
print(f"  Количество классов: {len(category_index)}")
print(f"  Классы: {[cat['name'] for cat in category_index.values()]}")

print("\n" + "="*60)
print("✅ ТЕСТИРОВАНИЕ ЗАВЕРШЕНО!")
print("="*60)
```

```
print("Исправляем отрицательные координаты...")

import tensorflow as tf
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import os

# 1. Создаем модель С ПРАВИЛЬНЫМИ ОГРАНИЧЕНИЯМИ
print("Создаем модель с ограниченными выходами...")

def create_bounded_detection_model():
    """Создает модель с ограниченными выходами (координаты в [0,1])"""
    # Базовая модель
    base_model = tf.keras.applications.MobileNetV2(
        input_shape=(640, 640, 3),
        include_top=False,
        weights='imagenet'
    )
    base_model.trainable = False

    inputs = tf.keras.Input(shape=(640, 640, 3))
    x = base_model(inputs, training=False)
    x = tf.keras.layers.GlobalAveragePooling2D()(x)
    x = tf.keras.layers.Dense(256, activation='relu')(x)

    # Выход для bounding box с сигмоидной активацией (ограничение [0,1])
    bbox_output = tf.keras.layers.Dense(4, activation='sigmoid', name='bbox')(x)

    # Выход для уверенности
    confidence_output = tf.keras.layers.Dense(1, activation='sigmoid', name='confidence')(x)

    model = tf.keras.Model(inputs=inputs, outputs=[bbox_output, confidence_output])
    return model

# Создаем исправленную модель
bounded_model = create_bounded_detection_model()
print("✅ Модель с ограничениями создана")

# 2. Загружаем веса (пробуем)
checkpoint_path = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/output/ckpt-3'
ckpt = tf.train.Checkpoint(model=bounded_model)

try:
    status = ckpt.restore(checkpoint_path)
    status.expect_partial()
    print("✅ Веса загружены")
except Exception as e:
    print(f"⚠️  Ошибка загрузки: {e}")

# 3. Тестируем на изображении
print("\nТестируем исправленную модель...")

image_path = '/content/drive/My Drive/tf_od_demo/my_data_2/my_data/test.jpg'
image = Image.open(image_path)
image_resized = image.resize((640, 640))
image_np = np.array(image_resized)

# Преобразуем и нормализуем
input_tensor = tf.convert_to_tensor(image_np, dtype=tf.float32)
input_tensor = input_tensor[tf.newaxis, ...] / 255.0

# Предсказание
bbox_pred, confidence_pred = bounded_model(input_tensor, training=False)

# Координаты теперь гарантированно в [0,1]
bbox = bbox_pred[0].numpy()
confidence = confidence_pred[0, 0].numpy()

print(f"Уверенность: {confidence:.1%}")
print(f"Координаты (нормализованные): {bbox}")

# 4. КОНВЕРТИРУЕМ КООРДИНАТЫ ПРАВИЛЬНО
print("\nПравильная конвертация координат...")

# Формат bounding box в детекции объектов обычно: [y_min, x_min, y_max, x_max]
# ИЛИ [x_min, y_min, x_max, y_max]

# Давайте попробуем оба варианта
def convert_bbox_format(bbox, format_type='yxyx'):
    """Конвертирует bounding box в разные форматы"""
    if format_type == 'yxyx':
        # [y_min, x_min, y_max, x_max]
        y_min, x_min, y_max, x_max = bbox
    elif format_type == 'xyxy':
        # [x_min, y_min, x_max, y_max]
        x_min, y_min, x_max, y_max = bbox
    elif format_type == 'cxcywh':
        # [center_x, center_y, width, height]
        center_x, center_y, width, height = bbox
        x_min = center_x - width/2
        x_max = center_x + width/2
        y_min = center_y - height/2
        y_max = center_y + height/2
    else:
        # Предполагаем [x_center, y_center, width, height]
        x_center, y_center, width, height = bbox
        x_min = x_center - width/2
        x_max = x_center + width/2
        y_min = y_center - height/2
        y_max = y_center + height/2

    return x_min, y_min, x_max, y_max

# Пробуем разные форматы
formats_to_try = ['yxyx', 'xyxy', 'cxcywh', 'default']

print("Пробуем разные форматы координат:")
for fmt in formats_to_try:
    try:
        x_min_norm, y_min_norm, x_max_norm, y_max_norm = convert_bbox_format(bbox, fmt)

        # Конвертируем в пиксели
        x_min_px = int(x_min_norm * 640)
        x_max_px = int(x_max_norm * 640)
        y_min_px = int(y_min_norm * 640)
        y_max_px = int(y_max_norm * 640)

        print(f"  {fmt}: [{x_min_px}, {y_min_px}, {x_max_px}, {y_max_px}]")

        # Проверяем на корректность (в пределах изображения)
        if (0 <= x_min_px < 640 and 0 <= x_max_px <= 640 and
            0 <= y_min_px < 640 and 0 <= y_max_px <= 640 and
            x_min_px < x_max_px and y_min_px < y_max_px):
            print(f"     ✅ Корректный bbox!")
            best_format = fmt
            best_coords = (x_min_px, y_min_px, x_max_px, y_max_px)
            break
    except:
        continue

# 5. АЛЬТЕРНАТИВНЫЙ ПОДХОД: Ищем объекты вручную на изображении
print("\nАнализируем изображение вручную...")

# Показываем изображение с сеткой
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

# Левая часть: оригинал с сеткой
axes[0].imshow(image_resized)
axes[0].set_title('Изображение с предполагаемыми областями')
axes[0].axis('off')

# Добавляем сетку
height, width = image_resized.size
grid_size = 80  # 8x8 сетка

for i in range(0, width, grid_size):
    axes[0].axvline(x=i, color='yellow', alpha=0.3, linestyle='--')
for i in range(0, height, grid_size):
    axes[0].axhline(y=i, color='yellow', alpha=0.3, linestyle='--')

# Правая часть: с предполагаемым bbox
axes[1].imshow(image_resized)

# Пробуем нарисовать bbox в разных местах чтобы найти объект
print("Ищем объекты визуально...")

# Определяем явные объекты на изображении (визуально)
# Создаем список предполагаемых объектов
potential_objects = []

# Область 1: предполагаемый основной объект (центр изображения)
obj1 = {
    'name': 'Основной объект',
    'coords': [width//4, height//4, 3*width//4, 3*height//4],  # Центральная область
    'color': 'red'
}

# Область 2: другие возможные объекты
obj2 = {
    'name': 'Второстепенный объект',
    'coords': [width//3, height//3, 2*width//3, 2*height//3],
    'color': 'green'
}

potential_objects.append(obj1)
potential_objects.append(obj2)

# Рисуем все предполагаемые объекты
for i, obj in enumerate(potential_objects):
    x_min, y_min, x_max, y_max = obj['coords']

    # Прямоугольник
    rect = plt.Rectangle((x_min, y_min), x_max-x_min, y_max-y_min,
                        fill=False, edgecolor=obj['color'], linewidth=2,
                        label=obj['name'])
    axes[1].add_patch(rect)

    # Текст
    axes[1].text(x_min, y_min-15, obj['name'],
                color='white', fontsize=10,
                bbox=dict(facecolor=obj['color'], alpha=0.8))

axes[1].legend(loc='upper right')
axes[1].set_title(f'Предполагаемые объекты (уверенность модели: {confidence:.1%})')
axes[1].axis('off')

plt.tight_layout()
plt.show()

# 6. ВЫВОДЫ И РЕКОМЕНДАЦИИ
print("\n" + "="*70)
print("ВЫВОДЫ И РЕКОМЕНДАЦИИ:")
print("="*70)

print(f"1. Уверенность модели: {confidence:.1%} - это хорошо!")
print(f"2. Координаты bbox: {bbox}")

if confidence > 0.5:
    print("3. ✅ Модель уверена в наличии объекта")
else:
    print("3. ⚠️  Уверенность низкая")

print("\n4. Проблемы:")
print("   • Модель обучена всего 3 шага (очень мало)")
print("   • Архитектура может не совпадать с checkpoint")
print("   • Bounding box выходит за границы изображения")

print("\n5. Что делать:")
print("   • ПРОДОЛЖИТЬ ОБУЧЕНИЕ до 1000+ шагов")
print("   • Использовать правильную архитектуру из конфига")
print("   • Проверить аннотации в CSV")

print("\n6. Ваши следующие шаги:")
print("   A. Обучить модель дольше:")
print("      !cd /content/models/research && python object_detection/model_main_tf2.py \\")
print("          --pipeline_config_path=\"/content/drive/My Drive/tf_od_demo/my_data_2/my_data/pipeline.config\" \\")
print("          --model_dir=\"/content/drive/My Drive/tf_od_demo/my_data_2/my_data/output\" \\")
print("          --num_train_steps=1000")

print("\n   B. Экспортировать правильно:")
print("      !cd /content/models/research && python object_detection/exporter_main_v2.py \\")
print("          --input_type image_tensor \\")
print("          --pipeline_config_path=\"/content/drive/My Drive/tf_od_demo/my_data_2/my_data/pipeline.config\" \\")
print("          --trained_checkpoint_dir=\"/content/drive/My Drive/tf_od_demo/my_data_2/my_data/output\" \\")
print("          --output_directory=\"/content/drive/My Drive/tf_od_demo/my_data_2/my_data/exported_final\"")

print("\n   C. Протестировать на других изображениях")
print("="*70)
```

```
#функция детекции для TF2
def detect_objects(image_np, model):
    """Детекция объектов на одном изображении"""
    #конвертируем в тензор TF
    input_tensor = tf.convert_to_tensor(image_np)
    #добавляем batch dimension (1, height, width, 3)
    input_tensor = input_tensor[tf.newaxis, ...]

    #выполняем детекцию
    detections = model(input_tensor)

    #обрабатываем результаты
    num_detections = int(detections.pop('num_detections'))

    #конвертируем в numpy
    result = {}
    for key, value in detections.items():
        result[key] = value[0, :num_detections].numpy()

    result['num_detections'] = num_detections
    result['detection_classes'] = result['detection_classes'].astype(np.int64)

    return result
```

```
#загружаем тестовое изображение
image_path = '/content/drive/My Drive/tf_od_demo/my_data/test.jpg'
print(f"\n3. Загружаем изображение: {image_path}")

if not os.path.exists(image_path):
    print(f" Файл не найден! Проверьте путь.")
    #что есть
    !ls -la "/content/drive/My Drive/tf_od_demo/my_data/"*.jpg 2>/dev/null
else:
    #загружаем изображение
    image_np = imageio.imread(image_path)
    print(f"Изображение загружено. Размер: {image_np.shape}")

    #запускаем детекцию

    output_dict = detect_objects(image_np, detect_fn)
    print(f"Детекция завершена. Всего обнаружений: {output_dict['num_detections']}")

    #пнализируем результаты

    confidence_threshold = 0.3

    #сколько уверенных детекций
    confident_mask = output_dict['detection_scores'] > confidence_threshold
    confident_count = np.sum(confident_mask)

    print(f" Порог уверенности: {confidence_threshold}")
    print(f" Уверенных детекций: {confident_count}")

    if confident_count > 0:
        print(f"Оценки уверенности: {output_dict['detection_scores'][confident_mask][:3]}")
    else:
        print("Нет уверенных детекций.Нужно уменьшить порог.")

    #визуализация

    #создаем копию изображения для рисования
    image_with_boxes = image_np.copy()

    #рисуем bounding boxes
    vis_util.visualize_boxes_and_labels_on_image_array(
        image_with_boxes,
        output_dict['detection_boxes'],
        output_dict['detection_classes'],
        output_dict['detection_scores'],
        category_index,
        use_normalized_coordinates=True,
        line_thickness=6,
        min_score_thresh=confidence_threshold)

    #отображаем результат
    plt.figure(figsize=(16, 12))
    plt.imshow(image_with_boxes)
    plt.axis('off')

    if confident_count > 0:
        plt.title(f'Обнаружено объектов: {confident_count} (порог: {confidence_threshold})',
                  fontsize=16, color='green')
    else:
        plt.title(f'Объекты не обнаружены (порог: {confidence_threshold})',
                  fontsize=16, color='red')

    plt.tight_layout()
    plt.show()

    import matplotlib
    matplotlib.use('Agg')
    %matplotlib inline


    plt.figure(figsize=(16, 12))
    plt.imshow(image_with_boxes)
    plt.axis('off')
    plt.title(f'Найдено. Уверенность: {output_dict["detection_scores"][confident_mask][0]:.1%}',
              fontsize=18, color='green')
    plt.show()
```

```
import matplotlib.pyplot as plt
from PIL import Image
import pandas as pd
import os

def show_grid(img_path):
    """Показать фото с сеткой 10px"""
    img = Image.open(img_path)
    plt.figure(figsize=(12, 9))
    plt.imshow(img)
    plt.grid(True, color='black', linewidth=1.2, alpha=0.4)
    plt.xticks(range(0, 641, 10))
    plt.yticks(range(0, 481, 10))
    plt.title('сетка 10px')
    plt.show()

def get_coords(img_name):
    """Получить координаты для двух объектов"""
    print(f"\n разметка фото: {img_name}")


    coords = {}

    print("\n МАТРЁШКА (doll):")
    coords['doll'] = [
        int(input("xmin (левая граница): ")),
        int(input("ymin (верхняя граница): ")),
        int(input("xmax (правая граница): ")),
        int(input("ymax (нижняя граница): "))
    ]

    print("\n ГНОМИК (dwarf):")
    coords['dwarf'] = [
        int(input("xmin (левая граница): ")),
        int(input("ymin (верхняя граница): ")),
        int(input("xmax (правая граница): ")),
        int(input("ymax (нижняя граница): "))
    ]

    return coords

def save_to_csv(img_name, coords, csv_path=None):
    """
    Сохранить координаты в CSV
    По умолчанию сохраняет на Drive
    """
    #если путь не указан, используем Google Drive
    if csv_path is None:
        csv_path = '/content/drive/My Drive/tf_od_demo/my_data_2/annotations_640x480.csv'

    #создаем папку если нет
    os.makedirs(os.path.dirname(csv_path), exist_ok=True)

    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
    else:
        df = pd.DataFrame(columns=['id','object_id','xmin','ymin','xmax','ymax','class_name','status'])

    #добавляем или обновляем строки
    new_data = [
        [img_name, 1, coords['doll'][0], coords['doll'][1], coords['doll'][2], coords['doll'][3], 'doll', 'Размечено'],
        [img_name, 2, coords['dwarf'][0], coords['dwarf'][1], coords['dwarf'][2], coords['dwarf'][3], 'dwarf', 'Размечено']
    ]

    #удаляем старые строки для этого фото если есть
    df = df[df['id'] != img_name]

    #добавляем новые
    for row in new_data:
        df.loc[len(df)] = row

    df.to_csv(csv_path, index=False)
    print(f" Координаты сохранены: {csv_path}")
    return csv_path
```

```
import pandas as pd
import os



DRIVE_CSV_PATH = '/content/drive/My Drive/tf_od_demo/my_data_2/annotations_640x480.csv'


try:
    img_name = current_photo_info['img_name']

    csv_path = DRIVE_CSV_PATH

    print(f"разметка фото: {img_name}")
    print(" "*50)

    #ввод координат
    print("\nМАТРЁШКА (doll):")
    doll_coords = [
        int(input("  xmin (левая граница): ")),
        int(input("  ymin (верхняя граница): ")),
        int(input("  xmax (правая граница): ")),
        int(input("  ymax (нижняя граница): "))
    ]

    print("\nГНОМИК (dwarf):")
    dwarf_coords = [
        int(input("  xmin (левая граница): ")),
        int(input("  ymin (верхняя граница): ")),
        int(input("  xmax (правая граница): ")),
        int(input("  ymax (нижняя граница): "))
    ]


    os.makedirs(os.path.dirname(csv_path), exist_ok=True)

    if os.path.exists(csv_path):
        df = pd.read_csv(csv_path)
    else:
        df = pd.DataFrame(columns=['id','object_id','xmin','ymin','xmax','ymax','class_name','status'])

    #обновляем существующие записи или добавляем новые:
    for i, row in df.iterrows():
        if row['id'] == img_name and row['object_id'] == 1:
            #обновляем матрёшку
            df.at[i, 'xmin'] = doll_coords[0]
            df.at[i, 'ymin'] = doll_coords[1]
            df.at[i, 'xmax'] = doll_coords[2]
            df.at[i, 'ymax'] = doll_coords[3]
            df.at[i, 'status'] = 'Размечено'
            break
    else:
        #если не нашли - добавляем новую запись
        df.loc[len(df)] = [img_name, 1, doll_coords[0], doll_coords[1], doll_coords[2], doll_coords[3], 'doll', 'Размечено']

    #обновляем гномика
    for i, row in df.iterrows():
        if row['id'] == img_name and row['object_id'] == 2:
            df.at[i, 'xmin'] = dwarf_coords[0]
            df.at[i, 'ymin'] = dwarf_coords[1]
            df.at[i, 'xmax'] = dwarf_coords[2]
            df.at[i, 'ymax'] = dwarf_coords[3]
            df.at[i, 'status'] = 'Размечено'
            break
    else:
        df.loc[len(df)] = [img_name, 2, dwarf_coords[0], dwarf_coords[1], dwarf_coords[2], dwarf_coords[3], 'dwarf', 'Размечено']



    #сохраняем
    temp_path = csv_path + '.temp'
    df.to_csv(temp_path, index=False)
    import shutil
    shutil.move(temp_path, csv_path)
    #результат

    print(f"Файл: {img_name}")
    print(f"CSV: {csv_path}")
    print("\nСохранённые координаты:")
    print(f"Матрёшка: {doll_coords}")
    print(f"Гномик: {dwarf_coords}")



except NameError:
    print("Ошибка: сначала запустите выбор фото")
    print(" Запустите ячейку с input('Введите номер фото')")
```
