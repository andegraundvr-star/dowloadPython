```python
import sympy as sp
```

```python
x, y = sp.symbols('x y')
```

```python
per = (2*x + 3*y)**2 - 3*x*(4/3*x + 4*y)
```

```python
print("Исходное выражение:")
print(per)
print()
```

```python

# Упрощаем выражение
per_simplified = sp.simplify(per)
print("Упрощенное выражение (simplify):")
print(per_simplified)
print()
```

```python
# Альтернативно: раскрываем скобки
per_expanded = sp.expand(per)
print("Раскрытое выражение (expand):")
print(per_expanded)
print()
```

```python
# Собираем выражение по переменным
per_collected = sp.collect(per, x)
print("Выражение, сгруппированное по x (collect):")
print(per_collected)
print()
```

```python
# Красивое отображение с помощью pretty print
print("Красивое отображение:")
sp.pprint(per_simplified)
print()
```

```python
import math
kor = (math.sqrt(13) + math.sqrt(7)) **2 / (10 + math.sqrt(91))
```

```python
# Упрощаем выражение
kor_simplified = sp.simplify(kor)
print("Упрощенное выражение (simplify):")
print(kor_simplified)
print()
```

```python
# Собираем выражение по переменным
kor_collected = sp.collect(kor, x)
print("Выражение, сгруппированное по x (collect):")
print(kor_collected)
print()
```

```python
# Альтернативно: раскрываем скобки
kor_expanded = sp.expand(kor)
print("Раскрытое выражение (expand):")
print(kor_expanded)
print()
```
