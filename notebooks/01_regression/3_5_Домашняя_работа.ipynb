Есть два набора точек - *x_hw* и *y_hw*. В рамках домашней работы нужно
* визуализировать набор точек
* найти коэффициенты регрессии $w_0, w_1$ по шагам, как в уроке
* посчитать предсказание в виде $\hat{y} = w_0 + w_1x$ и визуализировать его вместе с точками *x_hw* и *y_hw*

```python
x_hw = [50, 60, 70, 100]
y_hw = [10, 15, 40, 45]
```

```python
from matplotlib import pyplot as plt

plt.scatter(x_hw, y_hw, s=40, c='g', marker='o', alpha=0.9)
```

```python
#выводи матрицу X
import numpy as np
X = np.array([[1, 50],
        [1, 60],
        [1, 70],
        [1, 100]])
X
```

```python
#выводим вектор Y
Y = np.array([[10], [15], [40], [45]])
Y
```

```python
#для вывода к-ов регрессии сначала перемножаем матрицы 
w_analytic = (
    X.T.dot(X)
)
w_analytic
    
```

```python
from numpy.linalg import inv
#находим обратную матрицу 
w_analytic_inv = inv(w_analytic)
w_analytic_inv
```

```python
#перемножаем матрицы
w = w_analytic_inv.dot(X.T).dot(Y)
w
```

```python
#выводим результат перемножения - требуемые коэфициэнты
print(f"w0: {w[0][0]:.2f}, w1: {w[1][0]:.2f}")
```

```python
#считаем предсказание в виде $\hat{y} = w_0 + w_1x$
# задаём границы координатных осей
margin = 10
X_min = 20
X_max = X[:,1].max()+margin
# увеличиваем количество точек на прямой
X_support = np.linspace(X_min, X_max, num=100)
# предсказания нашей модели
Y_model = w[0][0] + w[1][0]*X_support
```

```python
# настройка графика 
plt.xlim(X_min, X_max)
plt.ylim(0, Y[:,0].max() + margin)
# рисуем исходные точки
plt.scatter(X[:,1], Y[:,0], s=40, c='g', marker='o', alpha=0.9)
# предсказания модели
plt.plot(X_support, Y_model)

plt.show()
```
