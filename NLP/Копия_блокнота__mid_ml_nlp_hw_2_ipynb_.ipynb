```python
import nltk
import string
import pandas as pd
from itertools import chain
import numpy as np



df = pd.read_csv('sample_data/brand_tweets_valid.csv', sep=',', encoding='utf8')
# удаляем строки, в которых отсутствует текст твита
df.drop(df[df.tweet_text.isnull()].index, inplace=True)
print(df.shape)

df.head()

```

```python
def tokenize_text(raw_text: str):
    """Функция для токенизации текста

    :param raw_text: исходная текстовая строка
    """
    tokenized_str = nltk.word_tokenize(raw_text)
    tokens = [i.lower() for i in tokenized_str if ( i not in string.punctuation )]
    filtered_tokens = [i for i in tokens if ( i not in stop_words )]
    return filtered_tokens

# применяем функцию в датафрейму с помощью метода .apply()
tokenized_tweets= df.tweet_text.apply(tokenize_text)

# добавляем новую колонку в исходный датафрейм
df = df.assign(
    tokenized=tokenized_tweets
)

df.tokenized.head()
```

```python

from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
list_lemmas = []
lemmatizer = WordNetLemmatizer()
for text in df['tweet_text']:
    tokens = word_tokenize(str(text))
    lemmas = [lemmatizer.lemmatize(token) for token in tokens]
    list_lemmas.append(lemmas)

print(f"Количество обработанных твитов: {len(list_lemmas)}")
print(f"Первый твит (токены): {list_lemmas[0][:10]}")
```

```python
# Собираем все токены в один список
all_tokens = []
for lemmas in list_lemmas:
    all_tokens.extend(lemmas)

# Количество уникальных токенов
unique_tokens = set(all_tokens)
print(f"Всего токенов: {len(all_tokens)}")
print(f"Уникальных токенов: {len(unique_tokens)}")
print(f"Примеры уникальных токенов: {list(unique_tokens)[:20]}")
```
